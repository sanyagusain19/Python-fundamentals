A set is the unordered collection of items. Every set element is unique (no duplicates)and must be immutable. However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical operations like union, intersection, symmetric difference etc.

```python
#empty set:
s={}
print(print(type(s)))
empty_set=set()
print(type(empty_set))

#1d:
s1={1,2,3}
print(s1)
#2d sets:
# s2={1,2,3,{3,4}}
# print(s2) #will throw error.
#hetero:
s3={1,2,'hello',True}
print(s3) # why only three values? where is true?
#can't have duplicates and python treats true as 1, that's why...

s4={1,2,3,'hello',(1,2,3),True } 
print(s4) #unordered result? - they are unordered : hashing is internally running deciding position.

#type conversion:
s4=set([1,2,3])
print(s4)
#set can't have mutable items
s5={1,2,[3,4]}
#it will throw an error.
```

```python
s={1,2,3}
set2={3,2,1}
print(s==set2)
```

```python
#Accessing items:
s1={1,2,3,4,5}
s1[0]
#dont follow indexing concept since unordered.
```

```python
#Editing items:
s1
s1[0]=100 #can't do this...
```

```python
#adding items:
s1
#add :
s1.add('python')
print(s1)
```

```python
#update: multiple items...
s1.update([5,6,7])
print(s1)
```

```python
#delete:
#del
set1={1,2}
del set1
#discard
set1={1,2,3}
set1.discard(2) #if value not present then it will not hrow an error.
print(set1)
#remove
sett={1,2,3,4,5}
sett.remove(4) #it will throw an error if value is not present
print(sett)
sett.pop()#random delete
#clear:
sett.clear() #make empty set
print(sett)
```

```python
#Set operations:
s1={1,2,3,4,5}
s2={4,5,6,7,8}
#union:
s1 | s2 #no dup
#intersection:
s1 &s2
#difference:
s1 - s2 #s1 elements not present in s2.
#symmetric difference:
s1 ^ s2 #except common other get printed
#membership test:
1 in s1 #true
#iteration:
for i in s1:
    print(i)
```

Set functions:

```python
# len/min/max/sum/sorted:
s={3,4,2,1,6}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s, reverse = True))
```

```python
#union/update:
s1
s2
#s1 | s2
s1.union(s2)

s1.update(s2) #permanent change in s1
print(s1)
print(s2)
```

```python
#intersection/diff update:
s1.intersection(s2)
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
s1.difference(s2)
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
#symmetric difference update:
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.symmetric_difference(s2)

s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
#is disjoint, is subset, is superset:
s1={1,2,3,4,5}
s2={4,5,6,7,8}
print(s1.isdisjoint(s2)) 
print(s1.issubset(s2))
print(s1.issuperset(s2))
```
