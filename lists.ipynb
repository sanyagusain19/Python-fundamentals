*TWO WAYS TO TRAVERSE LIST:*
  
  -ITEMWISE

  -INDEXWISE
  



```python
#ITEMWISE:
item= [1,2,3,4]
for i in item:
  print(i)
```

```python
#Indexwise:
item1=[2,3,4,5]
for i in range(0,len(item1)):
  print(f"istloop  {i}")
  print(f"2nd loop {item1[i]}")
```

ZIP()-Returns the first object, which is an iterator of tuple.In Python, zip() is a built-in function that combines elements from two or more iterables (like lists, tuples, strings, etc.) into tuples, pairing elements with the same index together.

```python
item
item2=[-1,-2,-3,-4]
zip_object=zip(item,item2)
print(zip_object)
print(list(zip_object))

res=[i+j for i,j in zip(item,item2)]
res
```
