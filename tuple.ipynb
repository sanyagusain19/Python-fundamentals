## Operations on tuples:


```python
#plus and multiplication
t1=(1,2,3,4)
t2=(5,6,7,8)
print(t1+t2)
print(t1*3)#3tuples formed
#membership
print(1 in t1)
print(6 not in t2)
#iteration
for i in t1:
    print(i)
```

## Functions in tuple:

```python
#len/sum/min/max/sorted:
t=(1,3,4,6,8,9)
print(len(t))
print(sum(t))
print(min(t))
print(max(t))
print(sorted(t , reverse=True))
```

```python
#count:
tup1=(11,3,3,6,7,5,6,9)
print(tup1.count(3))
print(tup1.count(1))
```

```python
#index:
tup1.index(6)#first occurence.
```

### tuples are faster than lists.

```python
import time
l1=list(range(10000))
tup2=tuple(range(10000))
start =time.time()
for i in l1:
    i*5
print(f"List time :{time.time()-start}")

start =time.time()
for i in tup2:
    i*5
print(f"tuple time :{time.time()-start}")
```

### It takes less memory as compared to list.


```python
import sys
print('list size:',sys.getsizeof(l1))
print("tuple size:",sys.getsizeof(tup2))
```
